Ein **Struktogramm** stellt den Ablauf eines Programms grafisch dar, unabhängig von der Programmiersprache. Es wird **von oben nach unten** gelesen. Bei einer **Fallauswahl** stehen die Bedingungen über den Spalten. Sie werden von links nach rechts geprüft, ausgeführt wird die erste Spalte, deren Bedingung zutrifft. Die Spalte **sonst** gilt, wenn keine Bedingung zutrifft.

## Run – Ausführen

Das folgende Programm setzt das Struktogramm in Python um. Führen Sie es dreimal aus und geben Sie nacheinander 15, 50 und 95 ein. Halten Sie den Abgleich mit Ihren Vorhersagen auf dem Arbeitsblatt fest.

````python
# Programm zum Struktogramm "Akkuladung"
ladestand = int(input("Ladestand in %: "))
if ladestand < 20:
    print("Sofort laden")
elif ladestand < 80:
    print("Laden möglich")
else:
    print("Ausreichend geladen")
print("Prüfung beendet")
````
