# Lösungen

## Predict und Run

Erwartet wird fast immer 8, tatsächlich erscheint `Die Summe ist: 53`. Das Programm läuft fehlerfrei und liefert trotzdem ein falsches Ergebnis.

## Investigate – Musterantworten

1. `print(type(zahl1))` liefert `<class 'str'>`. `input()` gibt immer Text zurück, auch bei Zifferneingabe.
2. Beide Operanden sind Text, also **verkettet** das `+` sie, statt zu addieren: aus `"5"` und `"3"` wird `"53"`. Der Operator richtet sich nach dem Datentyp der Operanden.
3. Ergebnisse:

| Ausdruck | Ergebnis | Bedeutung |
|---|---|---|
| `7 / 2` | `3.5` | Division, liefert immer eine Gleitkommazahl |
| `7 // 2` | `3` | Ganzzahldivision, Nachkommaanteil entfällt |
| `7 % 2` | `1` | Rest der Ganzzahldivision (Modulo) |
| `7 ** 2` | `49` | Potenz, hier 7 hoch 2 |

4. `type(7 / 2)` liefert `float`, `type(7 // 2)` liefert `int`. Auffällig: `/` liefert auch bei glatter Division eine Gleitkommazahl, `4 / 2` ergibt `2.0`.

## Modify – Schritt 1: Umwandlung, TypeError erscheint

Meldung: `can only concatenate str (not "int") to str`

````python
zahl1 = int(input("Erste Zahl: "))
zahl2 = int(input("Zweite Zahl: "))
summe = zahl1 + zahl2
print("Die Summe ist: " + summe)   # TypeError in dieser Zeile
````

## Modify – Schritt 2: Reparatur mit `str()`

````python
zahl1 = int(input("Erste Zahl: "))
zahl2 = int(input("Zweite Zahl: "))
summe = zahl1 + zahl2
print("Die Summe ist: " + str(summe))
````

## Modify – Schritt 3: dasselbe als f-String

````python
zahl1 = int(input("Erste Zahl: "))
zahl2 = int(input("Zweite Zahl: "))
summe = zahl1 + zahl2
print(f"Die Summe ist: {summe}")
````

## Modify – Schritt 4: Differenz und Produkt

````python
zahl1 = int(input("Erste Zahl: "))
zahl2 = int(input("Zweite Zahl: "))

print(f"Die Summe ist: {zahl1 + zahl2}")
print(f"Die Differenz ist: {zahl1 - zahl2}")
print(f"Das Produkt ist: {zahl1 * zahl2}")
````

## Make – Beispiellösung

````python
laenge = int(input("Rohrlänge in cm: "))
stueck = int(input("Stücklänge in cm: "))

anzahl = laenge // stueck
rest = laenge % stueck

print(f"Es entstehen {anzahl} Stücke, es bleiben {rest} cm Rest.")
````

## Erweiterung – Beispiellösung

Typischer Fehler: `rest // laenge` statt `/` – dann kommt immer 0 heraus.

````python
verschnitt = round(rest / laenge * 100, 1)
print(f"Verschnitt: {verschnitt} %")
````

## Schlüsselbegriffe – zum Nachschlagen und Lernen

| Begriff | Erklärung | Wie in Python umsetzen? |
|---|---|---|
| **Datentyp** | Art des Wertes, den eine Variable enthält | `int`, `float`, `str` |
| **type()** | zeigt an, welchen Datentyp ein Wert hat | `type(5)`, `type("5")` |
| **Typumwandlung** | einen Wert in einen anderen Datentyp überführen | `int("5")`, `str(8)`, `float("2.5")` |
| **Verkettung oder Addition** | das `+` verbindet Texte, addiert aber Zahlen | `"5" + "3"` ergibt `"53"`, `5 + 3` ergibt `8` |
| **f-String** | Text, in den Werte eingesetzt werden | `f"Summe: {summe}"` |
| **Ganzzahldivision und Rest** | `//` liefert die ganzen Anteile, `%` den Rest | `7 // 2` ergibt `3`, `7 % 2` ergibt `1` |

**Wichtige Hinweise zu Python**

- `input()` liefert immer einen Text, auch wenn Ziffern eingetippt werden.
- Die Division `/` liefert immer eine Gleitkommazahl: `4 / 2` ergibt `2.0`.
- `int("3.5")` scheitert. Bei Kommazahlen zuerst `float()` verwenden.
- Ein `TypeError` tritt auf, wenn Text und Zahl mit `+` verbunden werden sollen.
