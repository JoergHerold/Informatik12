## Aufgabe 3: Widerstand berechnen

Hier ist `float()` nötig und nicht `int()`, weil die Stromstärke in der Regel eine Kommazahl ist.

Mit U = 12 V und I = 0.5 A ergibt sich R = 12 / 0.5 = 24 Ohm.

Zwei Stolperstellen, die sich zum Besprechen eignen:

- Wer `0,5` mit Komma eingibt, bekommt einen `ValueError`. Python erwartet den Dezimalpunkt.
- Wer `int()` statt `float()` verwendet, bekommt bei `0.5` ebenfalls einen `ValueError`: `int("0.5")` scheitert, weil die Zeichenkette einen Punkt enthält.

````python
spannung = float(input("Spannung in Volt: "))
stromstaerke = float(input("Stromstärke in Ampere: "))

widerstand = round(spannung / stromstaerke, 2)

print(f"Der Widerstand beträgt {widerstand} Ohm")
````
