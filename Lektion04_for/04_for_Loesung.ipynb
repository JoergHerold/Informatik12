## Make – Erstellen

Fahrschulen rechnen mit Faustformeln (v in km/h, Wege in m):

| Größe | Faustformel |
|---|---|
| Reaktionsweg | v / 10 · 3 |
| Bremsweg | (v / 10)² |
| Anhalteweg | Reaktionsweg + Bremsweg |

Schreiben Sie ein Programm, das für die Geschwindigkeiten 30, 40, 50 und so weiter bis 130 km/h eine Tabelle ausgibt.

So könnte die Ausgabe beginnen:

```
Geschwindigkeit: Reaktionsweg + Bremsweg = Anhalteweg
30 km/h: 9.0 m + 9.0 m = 18.0 m
40 km/h: 12.0 m + 16.0 m = 28.0 m
...
```

*Hinweis:* Ein Quadrat berechnen Sie mit `** 2`.

### Testen und Kommentieren

Prüfen Sie Ihr Programm mit diesen Werten aus Ihrer Tabelle:

| Geschwindigkeit | erwarteter Anhalteweg |
|---|---|
| 30 km/h (erste Zeile) | 18.0 m |
| 50 km/h | 40.0 m |
| 100 km/h | 130.0 m |
| 130 km/h (letzte Zeile) | 208.0 m |

- Prüfen Sie besonders, ob die **erste und die letzte Zeile** erscheinen.
- Schreiben Sie hinter den Schleifenkopf einen **Kommentar**, welche Geschwindigkeiten die Schleife durchläuft und warum der Stoppwert so gewählt ist.

````python
print("Geschwindigkeit: Reaktionsweg + Bremsweg = Anhalteweg")
for v in range(30, 131, 10):    # 30 bis 130 km/h in 10er-Schritten, 131 ist der Stoppwert
    reaktionsweg = v / 10 * 3
    bremsweg = (v / 10) ** 2
    anhalteweg = reaktionsweg + bremsweg
    print(f"{v} km/h: {reaktionsweg} m + {bremsweg} m = {anhalteweg} m")
````

### Erweiterung

1. Hinter jede Zeile mit einem Anhalteweg über 100 m soll „Achtung“ erscheinen.
2. Geben Sie nach der Tabelle aus, bei wie vielen Geschwindigkeiten der Anhalteweg über 100 m liegt.

````python
print("Geschwindigkeit: Reaktionsweg + Bremsweg = Anhalteweg")
anzahl = 0
for v in range(30, 131, 10):    # 30 bis 130 km/h in 10er-Schritten
    reaktionsweg = v / 10 * 3
    bremsweg = (v / 10) ** 2
    anhalteweg = reaktionsweg + bremsweg
    zeile = f"{v} km/h: {reaktionsweg} m + {bremsweg} m = {anhalteweg} m"
    if anhalteweg > 100:
        zeile = zeile + "  Achtung"
        anzahl = anzahl + 1
    print(zeile)
print(f"Anhalteweg über 100 m bei {anzahl} Geschwindigkeiten")
````

### Zusatzaufgabe: Sichere Geschwindigkeit bei Nebel

Ermitteln Sie die höchste Geschwindigkeit, bei der man bei Nebel noch vor einem Hindernis in Sichtweite anhalten kann. Das Programm liest die Sichtweite in m ein und prüft die Geschwindigkeiten in 5-km/h-Schritten bis 200 km/h.

So könnte ein Programmlauf aussehen:

```
Sichtweite in m: 50
Höchste sichere Geschwindigkeit: 55 km/h
```

````python
sicht = float(input("Sichtweite in m: "))
sicher = 0
for v in range(5, 201, 5):
    anhalteweg = v / 10 * 3 + (v / 10) ** 2
    if anhalteweg <= sicht:
        sicher = v
print(f"Höchste sichere Geschwindigkeit: {sicher} km/h")
````
